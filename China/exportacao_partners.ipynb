```python
#importação de bibliotecas
import pandas as pd
```

```python
#ler o arquivo csv
caminho_arquivo = ('./downloadData_202503.csv')
dados = pd.read_csv(caminho_arquivo,delimiter=',')
```

```python
#cria dataframe

data_frame = pd.DataFrame(dados)
print(data_frame.head)
```

```python
#remove virgulas dos numeros
data_frame['US dollar'] = data_frame['US dollar'].astype(str).str.replace(',','', regex= False)
#converte de volta para número
data_frame['US dollar'] = pd.to_numeric(data_frame['US dollar'], errors = 'coerce')
```

```python
#cria coluna de ano
data_frame['ano']= data_frame['Date of data'].astype(str).str[:4]
```

```python
#cria coluna mês
data_frame['mes'] = data_frame['Date of data'].astype(str).str[4:6].str.zfill(2)
```

```python
#cria coluna dia
data_frame['dia'] = '01'
```

```python
#cria coluna data

data_frame['data'] = data_frame['dia'] + "/" + data_frame['mes'] + "/" + data_frame['ano']
```

```python
#ordena as colunas
data_frame = data_frame[['ano', 'mes', 'dia', 'data', 'Trading partner code', 'Trading partner', 'Commodity code','US dollar']]
```

```python
#cria quantidade estatística

data_frame['Quantidade Estatística'] = 0
```

```python
#converte colunas
data_frame = data_frame.astype({
    'ano': int,
    'mes': int,
    'dia': int
})
```

```python
#ordena os dados
data_frame = data_frame.sort_values(by=['mes','Commodity code'], ascending=True)
```

```python
#exporta para excel
data_frame.to_excel('./china.xlsx', index=False, header = True)
```
